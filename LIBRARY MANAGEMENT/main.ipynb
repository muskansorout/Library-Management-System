```python
import json
import random
import string
from pathlib import Path
from datetime import datetime, timedelta





class Library:

    database = "library.json"
    data = { "books" : [], "members" : [], "transactions" : []}

    if Path(database).exists():
        with open(database, "r") as f:
            data = json.load(f)

        data.setdefault("books", [])
        data.setdefault("members", [])
        data.setdefault("transactions", [])

    else:
        with open(database, "w") as f:
            json.dump(data, f, indent=4)

#####################################################################################

    @classmethod
    def save_data(cls):
        with open(cls.database ,"w") as f:
            json.dump(cls.data, f, indent=4, default = str)

#####################################################################################        

    def gen_id(Prefix = "B"):
        random_id = ""
        for i in range(5):
            random_id += random.choice(string.ascii_uppercase + string.digits)
        return Prefix + "-" + random_id
            

    

    def add_book(self):

        title = input("enter you book title: ")
        author = input("author of the book: ")
        copies = int(input("how many copies: "))

        book = {
            "book_id" : Library.gen_id(),
            "title" : title,
            "author" : author, 
            "copies" : copies,
            "available_copies" : copies,
            "added_on" : datetime.now().strftime("%Y-%m-%d %H:%M:%S")
        }
        
        Library.data["books"].append(book)
        print("-"*50)
        print("Book added successfully.")
        print("-"*50)

        Library.save_data()

#####################################################################################   

    def list_books(self):
        for book in Library.data["books"]:
            print("="*150)
            print(" Total Book")
            print("="*150)
            print(book)


#####################################################################################
    
    def gen_memid(Prefix = "M"):
        randomid = ""
        for i in range(5):
            randomid += random.choice(string.ascii_uppercase + string.digits)
        return Prefix + "-" + randomid
        

    def add_members(self):

        Member_name = input("Enter member name: ")

        members = {
            "Member_name" : Member_name,
            "Member_id" : Library.gen_memid(),
            "added_on" : datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
        }
        Library.data["members"].append(members)
        print("-"*50)
        print("members added successfully")
        print("-"*50)

        Library.save_data()


#####################################################################################
    
    def list_members(self):
        for members in Library.data["members"]:
            print(members)
#####################################################################################

    def borrow_book(self):
        # member
        
        member_name = input("enter name: ")
      

        member_found = False
        for member in Library.data["members"]:
            if member["Member_name"]  == member_name:
                member_found = True
                break

        if not member_found:
            print("member doesn't exist.")
            return

        #books
        
        book_name = input("enter book name.")
        book_found = False
        for book in Library.data["books"]:
            if book["title"] == book_name:
                book_found = True
                break
                
        if not book_found:
            print("book is not available.")
            return
            
            

        print("available copies: ", book["available_copies"])
              
        borrowed_copies = int(input("enter borrow borrowed copies: "))

        if book["available_copies"] < borrowed_copies:
            print("not available copies exist.")
            return
            
        print("="*50)
        book["available_copies"] -= borrowed_copies
        
        print("book borrowed successfully.","\nremaining book: ",book["available_copies"] ) 
        print("="*50)



        transaction =  self.create_transactions(
                  member_name,
                  book_name,
                  borrowed_copies
                  
        )

        Library.save_data()

#####################################################################################

    def gen_idT(Prefix = "T"):
        random_choice = ""
        for i in range(5):
            random_choice += random.choice(string.ascii_uppercase + string.digits)

        return Prefix + "-" + random_choice


    def create_transactions(self, member_name, book_name, borrowed_copies):

        
        issued_date = datetime.now()
        deadline =  issued_date + timedelta(days=14)
        
        transaction = {
            "transaction_id" : Library.gen_idT(),
            "member_name" : member_name,
            "book_name" : book_name,
            "borrowed_copies" : borrowed_copies,
            "issued_date" : datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
            "deadline" :  deadline.strftime("%Y-%m-%d %H:%M:%S"),
            "status": "borrowed",
            "return_date" : None
            

        }
        
        
        Library.data["transactions"].append(transaction)
        Library.save_data()
        return transaction


#####################################################################################

    
    def return_book(self):

        # memberfound
        
        member_namee = input("enter member name: ")

        member_foundd = False

        for member in Library.data["members"]:
            if member["Member_name"] == member_namee:
                member_foundd = True

                break

        if not member_foundd:
            print("member doesn't exist.")
            return


        # bookfound

        book_namee = input("enter book name: ")

        book_foundd = False

        for book in Library.data["books"]:
            if book["title"] == book_namee:
                book_foundd = True

                break

        if not book_foundd:
            print("book not found.")
            return


         # transactionfound   
       
        transaction_found = False
        for transaction in Library.data["transactions"]:

            if (transaction["member_name"] == member_namee
                 and transaction["book_name"] == book_namee
                 and transaction["borrowed_copies"] > 0):
            
                transaction_found = True
                break

        if not transaction_found:
                print("transaction doesn't exist")
                return      
                


        print("Available copies: ", book["available_copies"])

        return_books = int(input("how many books u want to return: "))

        if return_books > transaction["borrowed_copies"]:
            print("return book cant more than borrowed books.")
            return

        if book["available_copies"] + return_books > book["copies"]:
            print("Invalid Return!.")
            return

        book["available_copies"] += return_books
        print("book return successfully.")

        transaction["borrowed_copies"] -= return_books
        
        if transaction["borrowed_copies"] == 0:
            transaction["status"] ="returned"
            transaction["return_date"] = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
            
        Library.save_data()


#############################################################################################################


    def overdue_check(self):

                # memberfound
        
        member_namee = input("enter member name: ")

        member_foundd = False

        for member in Library.data["members"]:
            if member["Member_name"] == member_namee:
                member_foundd = True

                break

        if not member_foundd:
            print("member doesn't exist.")
            return


        # bookfound

        book_namee = input("enter book name: ")

        book_foundd = False

        for book in Library.data["books"]:
            if book["title"] == book_namee:
                book_foundd = True

                break

        if not book_foundd:
            print("book not found.")
            return



                 # transactionfound   
       
        transaction_found = False
        for transaction in Library.data["transactions"]:

            if (transaction["member_name"] == member_namee
                 and transaction["book_name"] == book_namee
                 and transaction["borrowed_copies"] > 0):
            
                transaction_found = True
                break

        if not transaction_found:
                print("transaction doesn't exist")
                return    




        if transaction["return_date"] is None:
            today = datetime.now()
            deadline = datetime.strptime(transaction["deadline"],"%Y-%m-%d %H:%M:%S")

            if today> deadline:
                print("book is overdue.")

            else:
                print("book is not overdue.")


        else:

            return_date = datetime.strptime(transaction["return_date"], "%Y-%m-%d %H:%M:%S")
            deadline = datetime.strptime(transaction["deadline"],"%Y-%m-%d %H:%M:%S")

            if return_date > deadline:
                print("book was returned late!.")

            else:
                print("book returned on time.")


             

#####################################################################################       

obj = Library()
#####################################################################################   
while True:
    print("="*50)
    print(" LIBRARY MANAGEMENT SYSTEM.")
    print("="*50)
    
    print("1.ADD BOOK: ")
    print("2.LIST BOOK: ")
    print("3.ADD MEMBERS: ")
    print("4.LIST MEMBERS ")
    print("5.BORROW BOOK: ")
    print("6.RETURN BOOK: ")
    print("7.OVERDUE_CHECK: ")
    print("0. EXIT: ")
    
    print("-"*50)
    
    choice = input("enter your choice: ")
    
    
    if choice == "1":
        obj.add_book()
   
    
    
    elif choice == "2":
        obj.list_books()
    
    elif choice == "3":
        obj.add_members()
    
    elif choice == "4":
        obj.list_members()
    
    
    elif choice == "5":
        obj.borrow_book()
    
    elif choice == "6":
        obj.return_book()
    
    elif choice == "7":
        obj.overdue_check()

    elif choice =="0":
        break





    
   

```
