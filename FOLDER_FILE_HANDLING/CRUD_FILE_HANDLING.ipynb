```python
from pathlib import Path
import shutil

# ******************* CREATE FOLDER *******************

def create_folder():
    try:
        
        name = input("enter your folder name: ")
        p = Path(name)
        if p.exists():                     # exists() : it checks if folder is already exist or not
            print("folder already exist.")
        else:
            
           p.mkdir()                       # mkdir() : it is a directory used to create  folder
           print("folder created successfully.")
    except Exception as err:
        print(f" sorry an error occurred as {err}.")
    

# ***************** READ FILES ND FOLDER ***********************


def read_file_folder():
    p = Path("")                     # path("") stnd for find the path of the folder (like jisme hm work kr rhe h {ye folder_nd_file ka  folder path find krega} )


    items = list(p.rglob('*'))       # rglob() is a method from Python's pathlib module that is used to recursively search for files and folders inside a directory and all of its subdirectories.

    for i,v in enumerate(items):     # enumerate:- 
       
          print(f"{i+1} : {v}")


# ********************* UPDATE FOLDER *********************

def update_folder():

    try:
        read_file_folder()
    
        old_name = input("enter your old folder name which want to chnge its name: ")
        p = Path(old_name)
        
        if p.exists() and p.is_dir():             # is.dir(): it is used to chck if directory of folder is exist or not
            new_name = input("enter your new folder name: ")
            new_p = Path(new_name)
            p.rename(new_p)
            print("folder name change successfully.")
            
        else:
            print("no such folder name exist.")



    except Exception as err:
        print(f" An error occurres as {err}.")


# ***************** DELETE FOLDER ****************************


def delete_folder():
    try:
        read_file_folder()
    
        name = input("enter your folder name which want ypu delete:-  ")
        p = Path(name)
    
        if p.exists() and p.is_dir():

            user = input("are u sure to delete this folder? y/n: ")

            if user.lower() == "y":
                shutil.rmtree(p)                             
                print("folder deleted successfully.")

            elif user.lower() == "n":
                print("folder is not deleted.")
                
            else:
                                            
              print("please enter y or n.")
                
        else:
            print("no such folder exist.")

    except Exception as err:
        print(f"an error occurred as {err}.")




# ************************* CREATE A FILE *************************

def create_file():
    try:
        read_file_folder()
    
        name = input("enter your file name: ")
        p = Path(name)
    
        if not p.exists():
            with open(name, 'w') as file:
                data = input("enter your input data what you want to write in a file: ")
                file.write(data)
                print("file created successfully.")
    
        else: 
            print("file already exist.")


    except Exception as err:
        print(f" an error occurred as {err}.")
        


# *********************** READ FILE *****************************

def read_file():
    try:
        name = input("enter your file name which want you to read: ")
        p = Path(name)
        if p.exists() and p.is_file():
            with open(name, 'r') as file:
                print(file.read())
    
    
        else:
            print("file doesn't exist")


    except Exception as err:
        print(f"an error occurred as {err}.")
        
    
# ******************* UPDATE FILE *************************

def update_file():
    try:
      
    
        name = input("enter your file name: ")
        p = Path(name)
    
        if p.exists() and p.is_file():
            with open(name, 'r') as file:
                print("\n -----context----.")
                print(file.read())
            with open(name, 'a' ) as file:
                user = input("\n enter your context here:- ")
                file.write("\n" + user)
                
            print("file updated successfully.")
            
        else:
            print("file doesn't exist")


    except Exception as err:
        print(f" an error occurred as {err}.")

       
# ******************** DELETE FILE *********************8

def delete_file():
    try:
        
        
        name = input("enter a file name which want you delete: ")
        p = Path(name)
        
        if p.exists() and p.is_file():
           

           with open(name, "r") as file:
               print("\n ------- file content --------")
               print(file.read())
            
           user= input("are you sure to delete your file  y/n : ")
           if user.lower() == 'y':
                p.unlink()
                print("file deleted successfully.")

           elif user.lower() == "n":
               print("file is not delted.") 
        
           else:
                print("please enter y or n")
        
        
        else:
            print("file doesn't found.")


    except Exception as err:
        print(f" an error occurred as {err}.")

        
        
        

        

# *********************** OPTIONS *****************************

    
print("options: ")
print("1. create a folder: ")
print("2. read files and folder.")
print("3. update the folder.")
print("4. delete the folder.")
print("5. create a file.")
print("6. read a file.")
print("7. update a file.")
print("8. delete a file.")



try:
    choice = int(input("enter your choice: "))

    if choice == 1:
        create_folder()
    
    elif choice == 2:
        read_file_folder()
    
    
    elif choice == 3:
        update_folder()
    
    
    elif choice == 4:
        delete_folder()
    
    elif choice == 5:
        create_file()

    elif choice == 6:
        read_file()

    elif choice == 7:
        update_file()

    elif choice == 8:
        delete_file()

    else:
        print("enter number from 1 to 8 only.")

except ValueError:
          print("\nENTER A VALID NUMBER\n.")



    
```
